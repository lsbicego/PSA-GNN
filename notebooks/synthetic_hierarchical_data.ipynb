# Synthetic hierarchical datasets

Generate and visualize examples from Macro-Hierarchies and Granular-Hierarchies. See the [README](../README.md#installation) for installation.

```python
from pathlib import Path
import sys

if not (Path.cwd() / "src").is_dir():
    sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
import networkx as nx

from src.data_generation.synthetic_datasets import (
    generate_hsbm_macro_topology_variable_dataset,
    generate_degree_matched_granularity_dataset_v2,
)
from src.datasets import SyntheticHierarchicalData
```

## Generate graphs

The generators return NetworkX graphs and their metadata. Adjust their parameters to create your own variants.

```python
macro, macro_meta = generate_hsbm_macro_topology_variable_dataset(num_graphs=2, seed=42)
granular, granular_meta = generate_degree_matched_granularity_dataset_v2(num_graphs=2, seed=42)

fig, axes = plt.subplots(2, 2, figsize=(9, 8))
for row, (name, graphs, metadata) in enumerate([
    ("Macro-Hierarchies", macro, macro_meta),
    ("Granular-Hierarchies", granular, granular_meta),
]):
    for col, (graph, meta) in enumerate(zip(graphs, metadata)):
        nx.draw(graph, nx.spring_layout(graph, seed=42), ax=axes[row, col],
                node_size=12, width=0.3, with_labels=False)
        axes[row, col].set_title(f"{name}, label={meta['label']}")
plt.tight_layout()
plt.show()
```
